```python
# SEPARAR EM CHUNKS OS ENVIOS PARA O BD

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=4000,
    chunk_overlap=0,
    length_function=len,
    add_start_index=True
)

chunks = text_splitter.split_documents(pages)
```

```python
# SALVAR NO VECTOR DB

db = Chroma.from_documents(
    chunks,
    embeddings_model,
    persist_directory="db"
)
```
